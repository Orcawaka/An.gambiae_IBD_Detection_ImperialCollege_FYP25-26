```python
import matplotlib.pyplot as plt
import numpy as np

x_bins = [0.5, 1.5, 2.5, 3.5, 4.5]

# Input power values 
r_default_power  = [0.5718, 0.0828, 0.0583, 0.0479, 0.0461] # Default
r_down_power = [0.0245, 0.0429, 0.0583, 0.0569, 0.0622] # r_low
r_up_power  = [0.8402, 0.0889, 0.1059, np.nan, np.nan] # r_high

def to_percent(val_list):
    return [v * 100 if not np.isnan(v) else np.nan for v in val_list]

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['figure.dpi'] = 300
plt.rcParams['axes.linewidth'] = 1.2

fig, ax = plt.subplots(figsize=(7.5, 5.5))

color_down = '#7f8c8d' 
color_def  = '#1f4e79'  
color_up   = '#e67e22' 

# Lower r
ax.plot(x_bins, to_percent(r_down_power), color=color_down, linestyle='-', 
        marker='o', markersize=7, linewidth=2.2, label='Lower $r$ ($1\\times10^{-9}$)')
# Default r
ax.plot(x_bins, to_percent(r_default_power), color=color_def, linestyle='-', 
        marker='s', markersize=7, linewidth=2.2, label='Default $r$ ($1\\times10^{-8}$)')
# Upper r
ax.plot(x_bins, to_percent(r_up_power), color=color_up, linestyle='--', 
        marker='^', markersize=8, linewidth=2.2, label='Higher $r$ ($5\\times10^{-8}$)*')

# Figure details
ax.set_ylabel('Proportion detected (Power %)', fontsize=12, fontweight='bold', labelpad=10)
ax.set_xlabel('Size of IBD segment (cM)', fontsize=12, fontweight='bold', labelpad=10)
ax.set_title('IBD Detection Power with IBDseq (r Comparison)', fontsize=13, pad=18, fontweight='bold')


ax.set_xticks(x_bins)
ax.set_xticklabels(['0-1', '1-2', '2-3', '3-4', '4-5'], fontsize=11)
ax.set_xlim(0, 5)
ax.set_ylim(0, 105)
ax.set_yticks([0, 20, 40, 60, 80, 100])

# Grid line
ax.grid(axis='y', linestyle='--', alpha=0.3)

# Legend
ax.legend(loc='upper right', frameon=True, facecolor='white', edgecolor='none', fontsize=10.5)

# Note
ax.text(0.0, -0.18, "*Note: For Higher $r$, total sequence length was restricted to 2.5 Mb ($2.5\\text{ cM}$),\n hence the curve terminates at the 2-3 cM bin due to simulation limits.", 
        fontsize=8.5, style='italic', color='#555555', transform=ax.transAxes, va='top', ha='left')
plt.subplots_adjust(bottom=0.18)

plt.tight_layout()

plt.savefig("../results/fyp_power_IBDSeq(Various r).png", format="png", bbox_inches='tight')

plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3])
r_labels = ['Lower $r$\n($1\\times10^{-9}$)', 'Default $r$\n($1\\times10^{-8}$)', 'Higher $r$\n($5\\times10^{-8}$)*']
# Input power values 
means = [99.20, 67.38, 61.97]
sds = np.array([5.57, 35.62, 34.94])

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['figure.dpi'] = 300
plt.rcParams['axes.linewidth'] = 1.2
fig, ax = plt.subplots(figsize=(7, 7))

ax.plot(x, means, color='#7f8c8d', linestyle='--', linewidth=2, zorder=1)
#Lower r:
ax.scatter(x[0], means[0], color='#7f8c8d', s=120, marker='o', 
           edgecolor='black', linewidth=1, label='Lower $r$', zorder=2)
#Default r:
ax.scatter(x[1], means[1], color='#1f4e79', s=120, marker='s', 
           edgecolor='black', linewidth=1, label='Default $r$', zorder=2)
# Higher r:
ax.scatter(x[2], means[2], color='#e67e22', s=140, marker='^', 
           edgecolor='black', linewidth=1, label='Higher $r$*', zorder=2)

ax.fill_between(x, 
                np.clip(means - sds, 0, 100), 
                np.clip(means + sds, 0, 100), 
                color='gray',       
                alpha=0.15,         
                zorder=0)
# Axis:
ax.set_ylabel('Mean Overlap Accuracy (%)', fontsize=12, fontweight='bold', labelpad=12)
ax.set_xlabel('Simulation Parameters (r Groups)', fontsize=12, fontweight='bold', labelpad=12)
ax.set_title('IBDSeq: Accuracy with various recombination rates', fontsize=13, pad=18, fontweight='bold', wrap = True)

ax.set_xticks(x)
ax.set_xticklabels(r_labels, fontsize=10.5)
ax.set_xlim(0.5, 3.5)
ax.set_ylim(0, 105)
ax.set_yticks([0, 20, 40, 60, 80, 100])
ax.grid(axis='y', linestyle='--', alpha=0.3)
ax.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='none', fontsize=10.5)
ax.text(0.0, -0.15, "*Note: Global Accuracy includes all valid evaluating segments across pairs.\nFor Higher $r$, evaluation is bounded within the restricted 2.5 Mb sequence limit.", 
        fontsize=8.5, style='italic', color='#555555', transform=ax.transAxes, va='top', ha='left')

plt.subplots_adjust(bottom=0.01)
plt.savefig("../results/fyp_accuracy_IBDSeq(Various r).png", format="png", bbox_inches='tight')

plt.show()
```
